```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
CHART_DIR = "charts"
os.makedirs(CHART_DIR, exist_ok=True)
sns.set_style("whitegrid")
plt.rcParams["figure.dpi"] = 120
```

```python
df=pd.read_csv("delhi_business_clean.csv")
print(f"\nLoaded {df.shape[0]} rows, {df.shape[1]} columns")
```

### Rating distribution by category (boxplot)

```python
plt.figure(figsize=(12, 6))
order = df.groupby("category_searched")["rating"].median().sort_values(ascending=False).index
sns.boxplot(data=df, x="category_searched", y="rating", order=order, palette="viridis")
plt.xticks(rotation=45, ha="right")
plt.title("Rating Distribution by Business Category", fontsize=14, fontweight="bold")
plt.xlabel("Category")
plt.ylabel("Rating")
plt.tight_layout()
plt.savefig(f"{CHART_DIR}/01_rating_by_category.png")
plt.close()
```

### Review Count vs Rating

```python
plt.figure(figsize=(9, 6))
sns.scatterplot(data=df, x="review_count", y="rating", hue="has_website", alpha=0.5, palette={0: "gray", 1: "seagreen"})
plt.xscale("log")
plt.title("Review Count vs Rating (log scale)", fontsize=14, fontweight="bold")
plt.xlabel("Number of Reviews (log scale)")
plt.ylabel("Rating")
plt.legend(title="Has Website", labels=["No", "Yes"])
plt.tight_layout()
plt.savefig(f"{CHART_DIR}/02_reviews_vs_rating.png")
plt.close()
```

### Business Count Heatmap (Area x Category)

```python
pivot_count = df.pivot_table(index="area", columns="category_searched",values="name", aggfunc="count", fill_value=0)
plt.figure(figsize=(16, 10))
sns.heatmap(pivot_count, cmap="YlGnBu", linewidths=0.3, annot=False)
plt.title("Business Density: Area x Category", fontsize=14, fontweight="bold")
plt.xlabel("Category")
plt.ylabel("Area")
plt.tight_layout()
plt.savefig(f"{CHART_DIR}/03_density_heatmap.png")
plt.close()
```

### Average rating heatmap - Area x Category

```python
pivot_rating = df.pivot_table(index="area", columns="category_searched",values="rating", aggfunc="mean")
plt.figure(figsize=(16, 10))
sns.heatmap(pivot_rating, cmap="RdYlGn", linewidths=0.3, vmin=3.5, vmax=5.0)
plt.title("Average Rating: Area x Category", fontsize=14, fontweight="bold")
plt.xlabel("Category")
plt.ylabel("Area")
plt.tight_layout()
plt.savefig(f"{CHART_DIR}/04_avg_rating_heatmap.png")
plt.close()
```

###  Top 10 areas by business count

```python
top_areas = df["area"].value_counts().head(10)
plt.figure(figsize=(10, 6))
sns.barplot(x=top_areas.values, y=top_areas.index, palette="crest")
plt.title("Top 10 Areas by Number of Businesses Listed", fontsize=14, fontweight="bold")
plt.xlabel("Number of Businesses")
plt.ylabel("Area")
plt.tight_layout()
plt.savefig(f"{CHART_DIR}/05_top_areas.png")
plt.close()
```

### Correlation matrix of numeric features

```python
numeric_cols = ["rating", "review_count", "has_website", "has_phone", "has_price_info", "num_types"]
plt.figure(figsize=(8, 6))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Matrix - Numeric Features", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.savefig(f"{CHART_DIR}/06_correlation_matrix.png")
plt.close()
```

```python
print(f"Overall average rating       : {df['rating'].mean():.2f}")
print(f"Category with highest rating : {df.groupby('category_searched')['rating'].mean().idxmax()}")
print(f"Category with lowest rating  : {df.groupby('category_searched')['rating'].mean().idxmin()}")
print(f"Avg rating WITH website      : {df[df['has_website']==1]['rating'].mean():.2f}")
print(f"Avg rating WITHOUT website    : {df[df['has_website']==0]['rating'].mean():.2f}")
print(f"Busiest area                 : {df['area'].value_counts().idxmax()}")
```
